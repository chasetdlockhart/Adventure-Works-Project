Cleaning quarter over quarter revenue growth.

```python
import pandas as pd

quarter_over_quarter_revenue_growth = pd.read_csv(r"C:\Users\chase\OneDrive\Desktop\Adventure Works Project\Raw csvs\quarter_over_quarter_revenue_growth.csv")
quarter_over_quarter_revenue_growth.info()
quarter_over_quarter_revenue_growth.head()
```

Must change "order_quarter" from str to datetime.

```python
quarter_over_quarter_revenue_growth["order_quarter"] = pd.to_datetime(quarter_over_quarter_revenue_growth["order_quarter"])

quarter_over_quarter_revenue_growth.info()
```

Creating the "is_measurable" column allows me to filter out null values that aren't measurable when creating Tableau visualizations.

```python
quarter_over_quarter_revenue_growth["is_measurable"] = quarter_over_quarter_revenue_growth["last_quarter_revenue"].notna()

quarter_over_quarter_revenue_growth["is_measurable"].value_counts()
```
